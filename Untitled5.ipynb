```python

# ============================================================
# CASE STUDY: E-COMMERCE CUSTOMER SEGMENTATION
# Algorithms: K-Means and DBSCAN
# Dataset: retail_sales_dataset.csv
# Techniques: RFM, StandardScaler, PCA and t-SNE
# ============================================================

# STEP 1: IMPORT LIBRARIES

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, DBSCAN
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from sklearn.metrics import silhouette_score

# STEP 2: LOAD DATASET

df = pd.read_csv("retail_sales_dataset.csv")

print("Dataset Shape:", df.shape)
print("\nFirst Five Rows:")
print(df.head())

print("\nMissing Values:")
print(df.isnull().sum())

# STEP 3: DATA PREPROCESSING

df["Date"] = pd.to_datetime(df["Date"])

# Remove rows with missing essential values
df = df.dropna(subset=[
    "Customer ID", "Date", "Total Amount"
])

# STEP 4: CALCULATE RFM VALUES

# Reference date = one day after the last transaction
reference_date = df["Date"].max() + pd.Timedelta(days=1)

rfm = df.groupby("Customer ID").agg({

    "Date": lambda x: (reference_date - x.max()).days,

    "Transaction ID": "count",

    "Total Amount": "sum"

})

# Rename columns
rfm.rename(columns={

    "Date": "Recency",
    "Transaction ID": "Frequency",
    "Total Amount": "Monetary"

}, inplace=True)

print("\nRFM Dataset:")
print(rfm.head())

# STEP 5: RFM INTERPRETATION

# Recency: Lower value means recently purchased
# Frequency: Higher value means purchased more often
# Monetary: Higher value means spent more money

print("\nRFM Statistics:")
print(rfm.describe())

# STEP 6: HANDLE SKEWNESS

# Log transformation reduces the effect of extreme values

rfm_log = np.log1p(rfm)

# STEP 7: FEATURE SCALING

scaler = StandardScaler()

X_scaled = scaler.fit_transform(rfm_log)

# Remove constant features after scaling.
# In this dataset Frequency is constant for all customers.

feature_mask = rfm_log.nunique() > 1

X = X_scaled[:, feature_mask]

print("\nFeatures used for clustering:")
print(rfm_log.columns[feature_mask].tolist())

# ============================================================
# STEP 8: FIND OPTIMAL NUMBER OF CLUSTERS USING SILHOUETTE
# ============================================================

silhouette_scores = []

for k in range(2, 9):

    kmeans_temp = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels_temp = kmeans_temp.fit_predict(X)

    score = silhouette_score(X, labels_temp)

    silhouette_scores.append(score)

    print("K =", k, "Silhouette Score =", round(score, 4))

# Visualize silhouette scores

plt.figure(figsize=(8, 5))

plt.plot(
    range(2, 9),
    silhouette_scores,
    marker="o"
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Finding Optimal K using Silhouette Score")
plt.show()

# Select K with highest silhouette score

optimal_k = range(2, 9)[np.argmax(silhouette_scores)]

print("\nOptimal K:", optimal_k)

# ============================================================
# STEP 9: APPLY K-MEANS CLUSTERING
# ============================================================

kmeans = KMeans(
    n_clusters=optimal_k,
    random_state=42,
    n_init=10
)

rfm["KMeans_Cluster"] = kmeans.fit_predict(X)

print("\nK-Means Cluster Distribution:")
print(rfm["KMeans_Cluster"].value_counts().sort_index())

# ============================================================
# STEP 10: APPLY DBSCAN CLUSTERING
# ============================================================

dbscan = DBSCAN(
    eps=0.8,
    min_samples=5
)

rfm["DBSCAN_Cluster"] = dbscan.fit_predict(X)

print("\nDBSCAN Cluster Distribution:")
print(rfm["DBSCAN_Cluster"].value_counts().sort_index())

# -1 represents noise or outliers

# DBSCAN Silhouette Score (only if at least 2 clusters exist)

db_labels = rfm["DBSCAN_Cluster"].values

valid_mask = db_labels != -1
valid_labels = db_labels[valid_mask]

if len(np.unique(valid_labels)) >= 2:

    db_score = silhouette_score(
        X[valid_mask],
        valid_labels
    )

    print("\nDBSCAN Silhouette Score:", round(db_score, 4))

else:
    print("\nDBSCAN did not find enough non-noise clusters.")

# ============================================================
# STEP 11: PCA DIMENSIONALITY REDUCTION
# ============================================================

pca = PCA(n_components=2)

X_pca = pca.fit_transform(X)

rfm["PCA1"] = X_pca[:, 0]
rfm["PCA2"] = X_pca[:, 1]

print("\nPCA Explained Variance:")
print(pca.explained_variance_ratio_)

print(
    "Total Variance Explained:",
    pca.explained_variance_ratio_.sum()
)

# ============================================================
# STEP 12: VISUALIZE K-MEANS USING PCA
# ============================================================

plt.figure(figsize=(9, 6))

sns.scatterplot(
    x="PCA1",
    y="PCA2",
    hue="KMeans_Cluster",
    palette="viridis",
    data=rfm,
    s=60
)

plt.title("K-Means Customer Segmentation using PCA")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.show()

# ============================================================
# STEP 13: VISUALIZE DBSCAN USING PCA
# ============================================================

plt.figure(figsize=(9, 6))

sns.scatterplot(
    x="PCA1",
    y="PCA2",
    hue="DBSCAN_Cluster",
    palette="tab10",
    data=rfm,
    s=60
)

plt.title("DBSCAN Customer Segmentation using PCA")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.show()

# ============================================================
# STEP 14: t-SNE DIMENSIONALITY REDUCTION
# ============================================================

tsne = TSNE(
    n_components=2,
    perplexity=30,
    random_state=42,
    init="pca",
    learning_rate="auto"
)

X_tsne = tsne.fit_transform(X)

rfm["TSNE1"] = X_tsne[:, 0]
rfm["TSNE2"] = X_tsne[:, 1]

# ============================================================
# STEP 15: VISUALIZE K-MEANS USING t-SNE
# ============================================================

plt.figure(figsize=(9, 6))

sns.scatterplot(
    x="TSNE1",
    y="TSNE2",
    hue="KMeans_Cluster",
    palette="viridis",
    data=rfm,
    s=60
)

plt.title("K-Means Segmentation using t-SNE")
plt.show()

# ============================================================
# STEP 16: VISUALIZE DBSCAN USING t-SNE
# ============================================================

plt.figure(figsize=(9, 6))

sns.scatterplot(
    x="TSNE1",
    y="TSNE2",
    hue="DBSCAN_Cluster",
    palette="tab10",
    data=rfm,
    s=60
)

plt.title("DBSCAN Segmentation using t-SNE")
plt.show()

# ============================================================
# STEP 17: CREATE CUSTOMER PERSONAS
# ============================================================

# Calculate average RFM values for each K-Means cluster

cluster_summary = rfm.groupby("KMeans_Cluster")[
    ["Recency", "Frequency", "Monetary"]
].mean().round(2)

print("\nK-Means Cluster Summary:")
print(cluster_summary)

# Calculate global median values

median_recency = rfm["Recency"].median()
median_monetary = rfm["Monetary"].median()

# Assign marketing personas

def create_persona(row):

    if row["Monetary"] >= median_monetary:

        if row["Recency"] <= median_recency:
            return "High-Value Active Customers"

        else:
            return "At-Risk High-Value Customers"

    else:

        if row["Recency"] <= median_recency:
            return "Recent Low-Spending Customers"

        else:
            return "Dormant Low-Value Customers"


cluster_summary["Persona"] = cluster_summary.apply(
    create_persona,
    axis=1
)

# Marketing strategies for each persona

marketing_actions = {

    "High-Value Active Customers":
    "Provide loyalty rewards, VIP membership and exclusive offers.",

    "At-Risk High-Value Customers":
    "Send personalized discounts, reminders and win-back campaigns.",

    "Recent Low-Spending Customers":
    "Offer product recommendations, bundles and cross-selling deals.",

    "Dormant Low-Value Customers":
    "Use re-engagement emails, limited-time coupons and low-cost promotions."

}

cluster_summary["Marketing Strategy"] = (
    cluster_summary["Persona"].map(marketing_actions)
)

print("\nCUSTOMER PERSONAS AND MARKETING STRATEGIES:")
print(cluster_summary.to_string())

# ============================================================
# STEP 18: CUSTOMER DEMOGRAPHIC AND CATEGORY PROFILE
# ============================================================

# Attach cluster labels to original customer records

customer_profile = df.merge(
    rfm[["KMeans_Cluster", "DBSCAN_Cluster"]],
    on="Customer ID",
    how="left"
)

print("\nAverage Age by K-Means Cluster:")

print(
    customer_profile.groupby("KMeans_Cluster")["Age"].mean()
)

print("\nGender Distribution:")

print(
    pd.crosstab(
        customer_profile["KMeans_Cluster"],
        customer_profile["Gender"],
        normalize="index"
    ).round(2)
)

print("\nPopular Product Categories by Cluster:")

for cluster in sorted(customer_profile["KMeans_Cluster"].unique()):

    print("\nCluster:", cluster)

    print(
        customer_profile[
            customer_profile["KMeans_Cluster"] == cluster
        ]["Product Category"].value_counts().head(3)
    )

# ============================================================
# STEP 19: FINAL COMPARISON
# ============================================================

print("\nFINAL CUSTOMER SEGMENTATION")

print("Total Customers:", len(rfm))

print("\nK-Means:")
print(rfm["KMeans_Cluster"].value_counts().sort_index())

print("\nDBSCAN:")
print(rfm["DBSCAN_Cluster"].value_counts().sort_index())

# ============================================================
# STEP 20: SAVE RESULTS
# ============================================================

rfm.to_csv("customer_segmentation_results.csv")

cluster_summary.to_csv("customer_personas.csv")

print("\nFiles saved successfully!")
```
